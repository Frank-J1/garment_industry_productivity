##  Data Cleaning

```python
#remove extra space in one of the 'finishing columns, combine the finishing columns, and fix 'sweing' typo
df.loc[df['department'] == 'finishing ', 'department'] = 'finishing'
df.loc[df['department'] == 'sweing', 'department'] = 'sewing'
df['department'].value_counts()
```

```python
# Removing the date column; the time frame is short and will most likely not be useful for our DTC
# Removing idle_men and idle_time and no_of_style_change because they are overwhelmingly 0
# Removing wip due to too many null values
df = df.drop(['date', 'idle_time', 'idle_men', 'wip', 'no_of_style_change'], axis=1, errors='ignore')

df.head()
```

```python
#Merge  quarter 5 with quater 4, we cannot have 5 week in a month and quater 5 cover very little values
df.loc[df['quarter'] == 'Quarter5', 'quarter'] = 'Quarter4'
df['quarter'].value_counts()
```

```python
# Change the values in the 'quarter' column to be numeric
# The column is originally a 'string' type, so we can simply just put the numeric value as a string and then convert
df.loc[df['quarter'] == 'Quarter1', 'quarter'] = '1'
df.loc[df['quarter'] == 'Quarter2', 'quarter'] = '2'
df.loc[df['quarter'] == 'Quarter3', 'quarter'] = '3'
df.loc[df['quarter'] == 'Quarter4', 'quarter'] = '4'
df['quarter'] = df['quarter'].astype('int64')
df['quarter'].value_counts()
```

```python
df[['quarter']].info()
```

```python
# 'number_of_workers' columns is currently a float, allowing for fractional workers
df['no_of_workers'] = df['no_of_workers'].astype('int64')
df.head(1)
```

```python
# Round 'actual_productivity' so it match 'targeted_productivity'
df['actual_productivity'] = df['actual_productivity'].round(2)
df.head(1)
```

```python
# Create a new column so we can use it in our DTC. Based on whether productivity was reached
df['productive'] = df['actual_productivity'] >= df['targeted_productivity']
df.head(20)
```

## Machine Learning Preperation
### Since department is only 2 values, we can make it a binary column, 0 for finishing and 1 for sewing
### Similarly, we cant use the values in quarter/day/team for our model, so we will make new columns for each quarter/day/team and have true/false to represent which quarter/day/team it is

```python
# Convert department column to binary values
df['department'] = df['department'].map({'finishing': 0, 'sewing': 1}).astype('int64')
df.head()
```

```python
df = pd.concat([df, pd.get_dummies(df['quarter'], prefix='q')], axis=1).drop(['quarter'], axis=1)
df.head()
```

df = pd.concat([df, pd.get_dummies(df['day'], prefix=None)], axis=1).drop(['day'], axis=1)
df.head()

```python
df = pd.concat([df, pd.get_dummies(df['team'], prefix='team')], axis=1).drop(['team'], axis=1)
df.head()
```

```python
df = pd.concat([df, pd.get_dummies(df['day'], prefix=None)], axis=1).drop(['day'], axis=1)
```

## Building the Decision Tree Classifier

```python
from sklearn.model_selection import train_test_split
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
```

```python
# Feature and target columns
X = df.drop(['actual_productivity', 'productive'], axis=1)
y = df['productive']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, shuffle=True, random_state=24)
```

```python
clf = tree.DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=24)
clf.fit(X_train, y_train)
```

```python
y_pred = clf.predict(X_test)
```

## Visualizing and evaluating teh tree

```python
from sklearn.metrics import accuracy_score
```

```python
print("Accuracy: ", round(accuracy_score(y_test, y_pred), 2))
```

```python
clf.classes_
```

```python
plt.figure(figsize=(22,10))
_ = plot_tree(clf,
              feature_names=X.columns,
              class_names=['Unproductive', 'Productive'],
              filled=True,
              rounded=True,
              proportion=True,
              fontsize=12)
            
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
print('Precision:', round(precision_score(y_test, y_pred), 2))
print('Recall:', round(recall_score(y_test, y_pred), 2))
print('F1 Score:', round(f1_score(y_test, y_pred), 2))
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(clf, X, y, cv= 10)
print('CV accuarcy scores:', scores.round(2))
print('Mean CV score:', scores.mean().round(2))
```

```python
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier(
    oob_score=True,
    random_state=24
)

forest.fit(X_train, y_train)
y_pred_forest = forest.predict(X_test)
print('Accuracy:', round(accuracy_score(y_test, y_pred_forest), 2))
```
